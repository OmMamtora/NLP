```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
temp_df = pd.read_csv("IMDB Dataset.csv")
temp_df.head()
```

```python
df = temp_df.iloc[:10000]
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

### Basic Preprocessing

```python
# Remove Tags
import re

def remove_tags(text):
    cleaned_text = re.sub(re.compile('<.*?>'), '', text)
    return cleaned_text
```

```python
df['review'] = df['review'].apply(remove_tags)
```

```python
df['review'][1]
```

```python
# Lower Case

df['review'] = df['review'].str.lower()
```

```python
df['review']
```

```python
# Remove Stopword
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def remove_Stopword(text):
    filter_review = []
    words = text.split()

    for review in words:
        if review.lower() not in stop_words:
            filter_review.append(review)
    return " ".join(filter_review)
```

```python
df['review'] = df['review'].apply(remove_Stopword)
```

```python
df['review']
```

```python
df
```

```python
X = df.iloc[:, 0:1]
y = df['sentiment']
```

```python
X
```

```python
y
```

### Machine Learning part

```python
# Label encoder
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
y
```

```python
# Train Test - Split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
X_train.shape
```

## Apply BoW 

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
```

```python
X_train_bow = cv.fit_transform(X_train['review']).toarray()
X_test_bow = cv.transform(X_test['review']).toarray()
```

```python
X_train_bow.shape
```

### Model

#### Model -1 

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix

gnb = GaussianNB()

gnb.fit(X_train_bow, y_train)
```

```python
y_pred = gnb.predict(X_test_bow)

accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

#### Model-2

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()

rf.fit(X_train_bow, y_train)

y_pred = rf.predict(X_test_bow)

accuracy_score(y_test, y_pred)
```

```python
cv1 = CountVectorizer(max_features=3000)

X_train_bow1 = cv1.fit_transform(X_train['review']).toarray()
X_test_bow1 = cv1.transform(X_test['review']).toarray()
```

```python
rf1 = RandomForestClassifier()

rf1.fit(X_train_bow1, y_train)
y_pred1 = rf1.predict(X_test_bow1)

accuracy_score(y_test, y_pred1)
```

### Using N-Grams

```python
cv2 = CountVectorizer(ngram_range=(1,3), max_features=7500)

X_train_bow2 = cv2.fit_transform(X_train['review']).toarray()
X_test_bow2 = cv2.transform(X_test['review']).toarray()

rf3 = RandomForestClassifier()
rf3.fit(X_train_bow2, y_train)
```

```python
y_pred2 = rf3.predict(X_test_bow2)
accuracy_score(y_test, y_pred2)
```

## Using TF-IDF
