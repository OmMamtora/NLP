```python
import pickle

import re
import string 
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
model =  pickle.load(open('spam_model.pkl', 'rb'))
vectorizer = pickle.load(open('tfidf_vectorizer.pkl', 'rb'))
```

```python
puncutation = string.punctuation
stop_word = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_data(text):

    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = text.translate(str.maketrans('', '',puncutation))

    words = text.split()
    words = [words for words in words if words not in stop_word]

    words = [lemmatizer.lemmatize(word) for word in words]
    return " ".join(words)
```

```python
def predict_email(text):
    text = clean_data(text)
    vectorized_text = vectorizer.transform([text])

    prediction = model.predict(vectorized_text)[0]
    # spam_prob = model.predict_proba(vectorized_text)[0][1]

    
    if prediction == 1:
        return f"Spam"
    else:
        return f"Not Spam"
```

```python
email = """
Congratulations!
You have won ₹50,000.
Click here to claim your prize now.
"""
print(predict_email(email))
```
