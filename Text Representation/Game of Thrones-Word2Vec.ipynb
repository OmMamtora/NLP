```python
import pandas as pd
import numpy as np

import gensim
import os
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
story = []

for filename in os.listdir('Dataset\Game of Thrones'):

    f = open(os.path.join('Dataset\Game of Thrones', filename), encoding='latin-1')

    corpus = f.read()
    raw_sent = sent_tokenize(corpus)


    for sent in raw_sent:
        words = simple_preprocess(sent)
        filter_story = [w for w in words if w not in stop_words]
        story.append(filter_story)
```

```python
len(story)
```

```python
story
```

```python
model = gensim.models.Word2Vec(
    window=10,
    min_count=2
)
```

```python
model.build_vocab(story) # Extract Unique word and build Vocabulary
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```
